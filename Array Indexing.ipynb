# Array Indexing

```python
import numpy as np
```

# Numpy Indexing & Slicing of Array

```python
arr=np.arange(11,21)
arr
```

```python
arr[4]
```

```python
#slicing means remove any pStopIteration

arr[1:4]

#for last element it is n+1
```

```python
arr[:5]
#default is first value
```

```python
arr[4:]
# default is last value
```

```python
arr[:]
```

```python
arr[3::2]

# here 2 is steps
```

```python
# we save it in to somewhere
arr_slice=arr[3::2]
arr_slice
```

# Numpy Indexing And Slicing of Matrix

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
# for rows
arr[0]
```

```python
arr[5]
```

```python
# for particulaar element

arr[0,0]
```

```python
arr[5,4]
```

```python
# slicing means remove particular portion 
# to slice 2,7,3,8

slice=arr[0:2,1:3]
slice
```

```python
slice=arr[3:,3:]
slice
```

```python
#or
slice=arr[3:6,3:5]
slice
```

```python
# for selecting column

arr[:,2]
```

# Boolean Indexing

```python
arr=np.arange(11,21)
arr
```

```python
# we check the number is even or not
bool_index =arr % 2==0
bool_index
```

```python
#now we assign boolean to array

arr=arr[bool_index]
arr
```
