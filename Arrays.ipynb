# Array Generation Function

```python
arr=np.arange(4,11,2)
arr

#arr ne array save hoto
```

```python
# to generate zeros

arr= np.zeros(6)
arr
```

```python
#now we create matrix
arr= np.zeros((6,6))
arr

# bracket under bracket is for matix or 2 dimensional array
```
