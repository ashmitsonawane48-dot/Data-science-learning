# Deep and shallow copy

```python
a=np.arange(1,21)
a
```

```python
slice=a[:5]
slice=slice*10
slice
```

```python
a
#there are no change in array this is the copy of shallow copy
```

```python
a[4:6]*10
```

```python
a
b=a
```

```python
b[0]=99
```

```python
b
```

```python
a
```

```python
# now there will be also change in a so this is deep copy 
```

```python
# so if we want that there will be no change in a so we use this method

a1=np.arange(1,21)
a1
```

```python
a1
b=a1.copy()
```

```python
b[0]=99
```

```python
b
```

```python
a1
```

```python
# so we see there will be no change in a1
```
