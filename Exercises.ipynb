```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
#over all sum

np.sum(s)
```

```python
# for sum of rows

np.sum(s,axis=1)

#1 because rows are horizantal
```

```python
# sum of columns

np.sum(s,axis=0)

# because colums are vertical
```

```python
import numpy as np

# Columns: [Math Marks, Science Marks]
data = np.array([
    [18, 75, 87],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
# get the shape of matrix
data.shape
```

```python
#find the average age of students 
np.mean(data[:,0])
```

```python
# etract math marks of all student
data[:,1]
```

```python
# find the highest science marks

np.max(data[:,2])
```

```python
# get details of student who scored more than 90

data[data[:,1]>90]
```

```python
data
```

```python
# increase math marks of all student by 5 

data[:,1] += 5
data
```

```python
# find how many students are younger than 19 

len(data[data[:,0]<19])
```

```python
# calculate the average marks in each subject (coluns wise)

np.mean(data[:,1:],axis=0)
```

```python
# get data of student who atleast score 80 in both subjectabs

data[(data[:,1]>=80) & (data[:,2] >=80)]
```

```python
# Replace all science marks<75 with 0 

data[:,2][data[:,2] <75] = 0
data
```
